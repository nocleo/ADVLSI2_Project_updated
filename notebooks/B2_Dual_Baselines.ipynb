# B2 — Dual classification baselines

This notebook launches the frozen B2 benchmark on Colab GPU. It runs the unchanged `NCSU_DRCNN` on `tile_random_reference` and `unseen_layout_v1` for seeds 42, 43, and 44, using 30 epochs, batch size 32, RMSprop, and train-only Manhattan augmentation. Results are checkpointed in Google Drive so a disconnected session can resume.

```python
from google.colab import drive
drive.mount('/content/drive')

from pathlib import Path
OUTPUT_DIR = Path('/content/drive/MyDrive/ADVLSI2_B2/b2_baselines')
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)
print(f'Persistent results: {OUTPUT_DIR}')
```

```python
import os
import shutil
import subprocess
import sys
import tempfile
from pathlib import Path

REPOSITORY = 'https://github.com/nocleo/ADVLSI2_Project_updated.git'
BRANCH = 'agent/b2-dual-classification-baselines'
CHECKOUT = Path(tempfile.mkdtemp(prefix='advlsi-b2-')) / 'ADVLSI2_Project_updated'

subprocess.run(
    ['git', 'clone', '--depth', '1', '--branch', BRANCH, REPOSITORY, str(CHECKOUT)],
    check=True,
)
os.chdir(CHECKOUT)

import torch
assert torch.cuda.is_available(), 'Select a GPU runtime in Colab before running B2.'
print('Commit:', subprocess.check_output(['git', 'rev-parse', 'HEAD'], text=True).strip())
print('PyTorch:', torch.__version__)
print('GPU:', torch.cuda.get_device_name(0))
```

```python
command = [
    sys.executable,
    'scripts/run_b2_benchmarks.py',
    '--python', sys.executable,
    '--output-dir', str(OUTPUT_DIR),
]
print(' '.join(command))
subprocess.run(command, check=True)
```
