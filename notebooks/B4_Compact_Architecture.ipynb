# B4 — Compact architecture experiment

This notebook compares the frozen B2 `NCSU_DRCNN` with one pre-registered `CompactBNPool` candidate. Candidate training uses only the unseen-layout training/validation split. Frozen-test evaluation and tile-random confirmation are unlocked only after the validation gate passes.
